### 4. Data Limits

1. **Unbalanced history:** Content records may have different amounts of historical data. Therefore, comparisons may be less reliable when historical coverage differs.

2. **GSC-only early rows:** Early records may contain Google Search Console metrics without complete engagement data. Search performance alone cannot establish overall content engagement.
3. **Window overlaps:** Metrics from different time windows may overlap. I will verify the actual window definitions before interpreting changes as independent trends.

4. **Missing data:** Several columns contain missing values, including `word_count`, `char_count`, `provider_used`, and `model_used`. Missing values should not automatically be interpreted as zero.

5. **Prediction limitations:** The dataset alone does not prove that a particular content change will cause better performance. Any prediction or ranking should be treated as an estimate, not a guarantee.
