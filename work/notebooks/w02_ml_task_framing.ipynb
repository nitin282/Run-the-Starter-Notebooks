## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```

A fixed rule like “if the message contains ‘free’, mark it as spam” is too simple for this data. Spam messages in spam_mail.csv use many different words, numbers and styles, and the same word can also appear in normal (ham) messages, so a single rule would flag real messages wrongly and miss new kinds of spam. The pattern depends on a combination of many signals, and writing an if-statement for every combination is not practical. A machine learning model can learn these patterns from the 5572 labelled messages and handle new messages better.
